Cleaning and Scaling the data

Author: Obiakor Joyce

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# Loading the data
train = pd.read_csv("train.csv", index_col=0)
test = pd.read_csv("test.csv", index_col=0)
X = train.drop('MEDV', axis=1)
y = train['MEDV']
X_test = test.copy()

# checking for outliers.
# checking Tax and Crime because those usually have huge numbers compared to the other columns.
plt.figure(figsize=(10, 4))
sns.boxplot(data=X[['CRIM', 'ZN', 'TAX']])
plt.title("Before Scaling: Notice how wide the TAX box is compared to the others")
plt.show()

# scaling the data TO ensure all the numbers are on a level playing field.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_test_scaled = scaler.transform(X_test)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# checking the boxes again after scaling.
plt.figure(figsize=(10, 4))
sns.boxplot(data=X_scaled_df[['CRIM', 'ZN', 'TAX']])
plt.title("After Scaling: Everything is centered around 0 now")
plt.show()
print(f"Old TAX average was {X['TAX'].mean():.2f}")
print(f"New scaled TAX average is {X_scaled_df['TAX'].mean():.2f}")
```
